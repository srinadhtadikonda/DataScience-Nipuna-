A **stem plot** is a technique used to classify either discrete or continous variables. A stem plot is used to organize data as they are collected. A stem plot looks something like a bar graph. Each number in the data is broken down into a stem. The stem of the number includes all but the last digit. The leaf of the number will always be a single digit.

```python
#Stem plot
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.stem(x,y)
plt.title("Stem plot")
plt.show()
```

```python
#stem plot
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0.1, 2*np.pi, 41)
y=np.exp(np.sin(x))
plt.stem(x,y)
plt.title("Stem plot")
plt.show()
```

Stacked bar plot stacks bars that represent different groups on top of each other. The height of the resulting bar shows the combined result of the groups.

```python
#Stacked bar plot
import matplotlib.pyplot as plt
import numpy as np

N = 5
boys = (20, 35, 30, 35, 27)
girls = (25, 32, 34, 20, 25)
boyStd = (2, 3, 4, 1, 2)
girlStd = (3, 5, 2, 3, 3)
ind = np.arange(N)
width = 0.35

fig = plt.figure(figsize=(10, 7))
p1 = plt.bar(ind, boys, width, yerr=boyStd)
p2 = plt.bar(ind, girls, width, bottom=boys, yerr=girlStd)

plt.ylabel('Contribution')
plt.title('Contribution by the teams')
plt.xticks(ind, ('T1', 'T2', 'T3', 'T4','T5'))
plt.yticks(np.arange(0, 81, 10))
plt.legend((p1[0], p2[0]), ('Boys', 'Girls'))

plt.show()
```

**Histogram**: A histogram is an accurate representation of the distribution of numerical data. It is an estimate of the probability distribution of a continuous variable. It is a kind of bar graph.

To construct a histogram follow these steps -

1) Bin the range of values

2) Divide the entire range of values into a series of intervals

3) Count how many values fall into each interval

The bins are usually specified as consecutiv, non-overlapping intervals if a variable.

```python
#Histogram
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(1,1)
a = np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
ax.hist(a, bins = [0,25,50,75,100])
ax.set_title("Histogram of result")
ax.set_xlabel("Marks")
ax.set_xticks([0,25,50,75,100])
ax.set_ylabel("No. of students")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import colors
from matplotlib.ticker import PercentFormatter


# Creating dataset
np.random.seed(23685752)
N_points = 10000
n_bins = 20

# Creating distribution
x = np.random.randn(N_points)
y = .8 ** x + np.random.randn(10000) + 25
legend = ['distribution']

# Creating histogram
fig, axs = plt.subplots(1, 1,
						figsize =(10, 7),
						tight_layout = True)


# Remove axes splines
for s in ['top', 'bottom', 'left', 'right']:
	axs.spines[s].set_visible(False)

# Remove x, y ticks
axs.xaxis.set_ticks_position('none')
axs.yaxis.set_ticks_position('none')

# Add padding between axes and labels
axs.xaxis.set_tick_params(pad = 5)
axs.yaxis.set_tick_params(pad = 10)

# Add x, y gridlines
axs.grid(visible=True, color ='grey',
		linestyle ='-.', linewidth = 0.5,
		alpha = 0.6)

# Add Text watermark
fig.text(0.9, 0.15, 'Jeeteshgavande30',
		fontsize = 12,
		color ='red',
		ha ='right',
		va ='bottom',
		alpha = 0.7)

# Creating histogram
N, bins, patches = axs.hist(x, bins = n_bins)

# Setting color
fracs = ((N**(1 / 5)) / N.max())
norm = colors.Normalize(fracs.min(), fracs.max())

for thisfrac, thispatch in zip(fracs, patches):
	color = plt.cm.viridis(norm(thisfrac))
	thispatch.set_facecolor(color)

# Adding extra features
plt.xlabel("X-axis")
plt.ylabel("y-axis")
plt.legend(legend)
plt.title('Customized histogram')

# Show plot
plt.show()
```

**Polar Plot:** Suppose thayt we have some data that is related to either direction or to some period cycle. It might make sense to plot that on a polar plot.

Example: Wind rose plot, used to show the strength of winds from each direction.

```python
#Polar plot(Circle)
import matplotlib.pyplot as plt
import numpy as np
plt.axes(projection = 'polar')
r = 2  #Setting the radius
rads = np.arange(0, (2 * np.pi), 0.01)
for rad in rads:
  plt.polar(rad, r, 'g.')

plt.show()
```

```python
#Polar plot(Ellipse)
import numpy as np
import matplotlib.pyplot as plt
import math

#setting the axes projection as polar
plt.axes(projection = 'polar')

#setting the values of semi-major and semi-minor axes

a=4
b=3

#Creating an array containing the radian values
rads = np.arange(0, (2 * np.pi), 0.01)

#plotting the ellipse
for rad in rads:
  r = (a*b)/math.sqrt((a*np.sin(rad))**2 + (b*np.cos(rad))**2)
  plt.polar(rad, r, 'g.')

plt.show()
```

```python
#Polar plot(Cardiod)
import numpy as np
import matplotlib.pyplot as plt
import math

plt.axes(projection = 'polar')

#setting the length of axis of cardiod
a=4

#creating an array containing the radian values
rads = np.arange(0, (2 * np.pi), 0.01)

#plotting the cardioid
for rad in rads:
  r = a*np.cos(rad)
  plt.polar(rad, r, 'g.')

plt.show()
```

```python
#Polar plot(Archimedian spiral): An archimedean spiral is the locus of a point moving uniformly on a straight line, which itself is turning
#unifromLy about one of its end points. Here r is tdefined as r = theta

import numpy as np
import matplotlib.pyplot as plt

plt.axes(projection = 'polar')

#creating an array comtating the radian values
rads = np.arange(0, 2 * np.pi, 0.01)

#plotting the archimedean spiral
for rad in rads:
  r = rad
  plt.polar(rad, r, 'g.')

plt.show()
```

```python
#Polar plot(Rhodonea): A Rhodonea or Rose curve is a rose-shaped sinusoid plotted in polar coordinates. Here, r is defined as : r = =acos(ntheta)
# a = length of petals, n= number of petals

import numpy as np
import matplotlib.pyplot as plt

plt.axes(projection = 'polar')

a=1
n=6

#creating an array comtating the radian values
rads = np.arange(0, 2 * np.pi, 0.01)

#plotting the rose
for rad in rads:
  r = a * np.cos(n*rad)
  plt.polar(rad, r, 'g.')

plt.show()
```

```python
#pie chart
import matplotlib.pyplot as plt
import numpy as np

cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [23,17,35,29,12,41]

fig = plt.figure(figsize = (10, 7))
plt.pie(data, labels = cars)
plt.title("Pie chart")
```

```python
#Pie chart with explode
import matplotlib.pyplot as plt
import numpy as np

cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [23,17,35,29,12,41]

explode = (0.1,0.0,0.2,0.3,0.0,0.0)

colors = ("orange", "cyan", "brown", "grey", "indigo", "beige")

wp = {'linewidth': 1, 'edgecolor': "green"}

def func(pct, allvalues):
    absolute = int(pct / 100.*np.sum(allvalues))
    return "{:.1f}%\n({:d} g)".format(pct, absolute)

fig, ax = plt.subplots(figsize=(10,7))
wedges, texts, autotexts = ax.pie(data,
                                  autopct=lambda pct: func(pct,data),
                                  explode=explode,
                                  labels=cars,
                                  shadow=True,
                                  colors=colors,
                                  startangle=90,
                                  wedgeprops=wp,
                                  textprops=dict(color="magenta"))

ax.legend(wedges, cars,
          title = "Cars",
          loc = "center left",
          bbox_to_anchor = (1, 0, 0.5, 1))

plt.setp(autotexts, size = 8, weight = "bold")
ax.set_title("Customizing pie chart")

plt.show()
```

Error bars help you indicate estimated error or uncertainity to give a general sense of how precise a measurement is this is done through the use of markers drawn over the original graph and its data points.

```python
#Error bar
import matplotlib.pyplot as plt
#making a simple plot
x = [1,2,3,4,5,6,7]
y = [1,2,1,2,1,2,1]
y_error = 0.2
plt.plot(x,y)
plt.errorbar(x,y,yerr=y_error,fmt='o')
```

**Violin Plot:** A violin plot is used to represent the distribution of a dataset. It combines elements of a box plot and a kernel density plot. The plot consists of symmetrical shapes that look like violins to represent different groups. The "width" of these shapes represnets the density of the data at that value. Wider sections indicate higher density i.e. more data points fall within that range, while narrower section indicates lower density.

```python
#violin plot
import matplotlib.pyplot as plt
import numpy as np

# Generate data with different standard devations
data = [np.random.normal(0, std, 100) for std in range(1, 5)]

# Create violin plot
plt.violinplot(data, showmeans = False, showmedians = True)
plt.title("Basic violin plot")
plt.show()
```

```python
#violin plot
import matplotlib.pyplot as plt
import numpy as np

# Generate data with different standard devations
data = [np.random.normal(0, std, 100) for std in range(1, 5)]

# Create violin plot
plt.violinplot(data, showmeans = False, showmedians = True, vert = False)
plt.title("Horizontal plot")
plt.show()
```

**Area Plot:** An area chart is similar to a line chart, except that the area between the x axis and the line is filled in with the color or shading. It represnts the evolution of a numerci variable.

```python
#Area plot
import matplotlib.pyplot as pyplot

days = [1,2,3,4,5]
age = [63,81,52,22,37]
weight = [17,28,72,52,32]

pyplot.plot([],[],color='c',label='Weather predicted', linewidth = 5)
pyplot.plot([],[],color='g', label = 'Weather change happened', linewidth = 5)
pyplot.stackplot(days, age, weight, colors=['c','g'])
pyplot.xlabel('Fluctuation  with time')
pyplot.ylabel('Days')
pyplot.title('Weather report using area plot')
pyplot.legend()
pyplot.show()
```

**Subplots**: The matplotlib.pyplot.subplots method provides a way to plot multiple plots on a single figure.

```python
#Subplots in matplotlib using 1D array of subplots
import matplotlib.pyplot as plt

x = [1,2,3]
y = [0,1,0]
z = [1,0,1]

fig,ax = plt.subplots(2)
ax[0].plot(x,y)
ax[1].plot(x,z)
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0.0, 2.0, 0.01)
y = 1 + np.sin(2 * np.pi * x)

fig, ((ax1, ax2), (ax3, ax4), (ax5,ax6)) = plt.subplots(3, 2)
ax1.plot(x, y, color = "orange")
ax2.plot(x, y, color = "green")
ax3.plot(x, y, color = "blue")
ax4.plot(x, y, color = "magenta")
ax5.plot(x, y, color = "black")
ax6.plot(x, y, color = "red")
```

```python
#Subplots with titles
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])
fig, ax = plt.subplots(2,2)

ax[0,0].plot(x,x)
ax[0,1].plot(x,x**2)
ax[1,0].plot(x,x*x)
ax[1,1].plot(x,x*x*x)
ax[0,0].set_title("Linear")
ax[0,1].set_title("Double")
ax[1,0].set_title("Square")
ax[1,1].set_title("Cube")
fig.tight_layout()
plt.show()
```

```python
#Subplots with titles using loops

import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])
fig, ax = plt.subplots(2,2)
title = ["Linear", "Double", "Square", "Cube"]
y = [x, x**2, x*x, x*x*x]

for i in range(4):
  plt.subplot(2,2,i+1)
  plt.plot(x,y[i])
  plt.gca().set_title(title[i])

plt.tight_layout()
plt.show
```
