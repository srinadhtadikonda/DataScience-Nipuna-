```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb

def sinplot(flip=1):
    x = np.linspace(0,14,100)
    for i in range(1,5):
        plt.plot(x, np.sin(x+i*.5)*(7-i)*flip)

sb.set()
sinplot()
plt.show()
```

**Strip plot:** stripplot() is used when one of the variable under study is categorical. It represents the data in sorted order along any one of the axis.

```python
#stripplot
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("iris")
sb.stripplot(x="species", y="petal_length", data = df)
plt.show()
```

```python
#strip plot with jitter
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("iris")
sb.stripplot(x="species", y="petal_length", data = df, jitter = True)
plt.show()
```

**Swarm plot:** Another option which can be used as an alternative to 'Jitter' is function swarmplot(). This function positions each point of scatter plot on the categorical axis and thereby avoids overlapping points.

```python
#swarmplot
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("iris")
sb.swarmplot(x="species", y="petal_length", data = df)
plt.show()
```

**Statistical estimation:** In most of the situations, we deal with estimations of the whole distribution of the data. But when it comes to cental tendency estimation, we need specific way to summarize the distribution. Mean and median are the very often used techniques to estimate central tendency of the distribution. Plots with which we acn estimate the central tendency of the distribution

```python
#Statistical estimation
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("titanic")
sb.barplot(x="sex", y="survived", hue = "class", data = df)
plt.show()
```

```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("titanic")
sb.countplot(x="class", data = df, hue = "class", palette = "Blues")
plt.show()
```

**Point plot:** Point plots serve same as the bar plots but in a different style. Rather than the full bar, the value of the estimate is represented by the point as a certain height on the other axis.

```python
#Point plot
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("titanic")
sb.pointplot(x="sex", y="survived", hue = "class", data = df)
plt.show()
```

**Wide form data:** It is always preferable to use 'long-from' or 'tidy' datasets. Bu at times when we are left with no option rather than to use a 'wide-form' dataset, same functions can also be applied to "wide-form" data in a variety of formats, including pandas data frames or 2D NumPy arrays. These objects should be passed directly to the data parameter the x and y variables must be specified as strings.

```python
#Boxplot
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("iris")
sb.boxplot(data = df, orient = "h")
plt.show()
```

**Factor plot:** Factor plot draws a categorical plot on a FacetGrid. Using 'kind' parameter we can choose the plot like boxplot, violinplot, barplot, stripplot. FacetGrid uses pointplot by default.

```python
#Factorplot not supported hence we use catplot
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("exercise")
sb.catplot(x="time", y="pulse", hue = "kind", data = df)
plt.show()
```

```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df = sb.load_dataset("exercise")
sb.catplot(x="time", y="pulse", hue = "kind", kind = "violin", data = df)
plt.show()
```
