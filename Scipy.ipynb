The scipy library depends on NumPy, which provides convenient and fast N- dimensional array manipulation. The SciPy library is built to work with NumPy and provides many user-friendly and efficient numerical practices such as routines for numerical integration and optimization.

**Floyd Warshall**: To find the shortest path between all pairs of elements.

```python
#floyd_warshall
import numpy as np
from scipy.sparse.csgraph import floyd_warshall
from scipy.sparse import csr_matrix
arr=np.array([[0,1,2],[1,0,0],[2,0,0]])
newarr=csr_matrix(arr)
print(floyd_warshall(newarr,return_predecessors=True))
```

**Bellaman Ford:** Find the shortest path between all pairs of elements, but this method can handle negative weights as well.

```python
#bellman_ford
import numpy as np
from scipy.sparse.csgraph import bellman_ford
from scipy.sparse import csr_matrix
arr=np.array([[0,-1,2],[1,0,0],[2,0,0]])
newarr=csr_matrix(arr)
print(bellman_ford(newarr,return_predecessors=True,indices=0))
```

**T-Test**: T-tests are used to determine if there is a significant deference between means of two varibles and lets us know if they belong to the same distribution. It is a two tailed test

```python
#ttest
import numpy as np
from scipy.stats import ttest_ind
v1=np.random.normal(size=100)
print(v1)
print("--------------------------")
v2=np.random.normal(size=100)
print(v2)
print("--------------------------")
res=ttest_ind(v1,v2)
print(res)
print("--------------------------")
res=ttest_ind(v1,v2).pvalue
print(res)
```

**KS test**: KS test is used to check if given values follow a distribution. The function takes the value to be tested, and the CDF as two parameters. A CDF can be either a string or a callable function that returns probability. It can be one tailed or two tailed test.

```python
#kstest
import numpy as np
from scipy.stats import kstest
v=np.random.normal(size=100)
res=kstest(v,'norm')
print(res)
```

**Scipy Optimize:** The optimize package provides various commonly used optimization algorithms.

```python
import numpy as np
from scipy.optimize import minimize

def rosen(x):
  return sum(100.0*(x[1:]-x[:-1]**2.0+(1-x[:-1]**2.0)))

x0=np.array([1.3,0.7,0.8,1.9,1.2])
res=minimize(rosen,x0,method="nelder-mead")
print(res.x)
```

**Linear regression**: Linear regression is used to find the relationship between the two variables. The scipy proveides linregress() function to perform linear regression.

```python
#simple linear regression
import numpy as np
from scipy import stats

x=np.array([1,2,3,4,5])
y=np.array([2,3.9,6.1,8.2,9.8])
slope,intercept,r_value,p_value,std_err=stats.linregress(x,y)
print("Slope:",slope)
print("Intercept:",intercept)
print("R squared",r_value**2)
print("p-value",p_value)
print("Standard error",std_err)
```

```python
#Multiple linear regression
import numpy as np
import statsmodels.api as sm
from scipy import stats

x1=np.array([1,2,3,4,5])
x2=np.array([2,4,6,8,10])
y=np.array([3,5,7,9,11])

X=np.column_stack((x1,x2))
X=sm.add_constant(X)

model=sm.OLS(y,X)
result=model.fit()

print("Slope coefficients:",result.params[1:])
print("SIntercept coefficients:",result.params[0])
print("R-squared value:",result.rsquared)
```

Spectogram: The signal processing toolbox consists of few filtering functions,a limited set of filter design tools. It also contains few B-spline interpolation algorithm for one- and two- dimensional data.

```python
from scipy import signal
import matplotlib.pyplot as plt
import numpy as np

fs=10e3
N=1e5
amp=2*np.sqrt(2)
noise_power=0.001*fs/2
time=np.arange(N)/fs
freq=np.linspace(1e3,2e3,int(N))

x=amp*np.sin(2*np.pi*freq*time)
x+=np.random.normal(scale=np.sqrt(noise_power),size=time.shape)

f,t,Sxx=signal.spectrogram(x,fs)
plt.pcolormesh(t,f,Sxx)
plt.ylabel('Frequency [Hz]')
plt.xlabel('Time [sec]')
plt.show()
```
